````python
# Import required libraries
import os
import sys
import subprocess
import time
import threading
from pathlib import Path

# Add project root to path
project_root = Path().absolute().parent
sys.path.append(str(project_root))

print(f"Project root: {project_root}")
print(f"Current working directory: {os.getcwd()}")
````

## 1. 🚀 Streamlit Application Setup

Our Streamlit app is already created in the `ui/app.py` file. Let's check if all required packages are installed.

````python
# Check if required packages are installed
required_packages = [
    'streamlit',
    'plotly',
    'pyngrok',
    'pandas',
    'numpy',
    'scikit-learn',
    'joblib'
]

missing_packages = []

for package in required_packages:
    try:
        __import__(package.replace('-', '_'))
        print(f"✅ {package} is installed")
    except ImportError:
        print(f"❌ {package} is missing")
        missing_packages.append(package)

if missing_packages:
    print(f"\n📦 Missing packages: {missing_packages}")
    print("Please install them using: pip install " + " ".join(missing_packages))
else:
    print("\n🎉 All required packages are installed!")
````

## 2. 🔧 Application Configuration

````python
# Check if our Streamlit app file exists and show its structure
app_file = project_root / 'ui' / 'app.py'

if app_file.exists():
    print(f"✅ Streamlit app found: {app_file}")
    
    # Get file size
    file_size = app_file.stat().st_size
    print(f"📊 App file size: {file_size} bytes")
    
    # Count lines
    with open(app_file, 'r') as f:
        lines = f.readlines()
    print(f"📝 Total lines of code: {len(lines)}")
    
    # Show first few lines
    print("\n📋 First 10 lines of the app:")
    for i, line in enumerate(lines[:10], 1):
        print(f"{i:2d}: {line.rstrip()}")
        
else:
    print(f"❌ Streamlit app not found: {app_file}")
    print("Please make sure the app.py file is created in the ui/ directory")
````

## 3. 🧪 Local Testing

Before deploying publicly, let's test the Streamlit app locally.

````python
# Function to run Streamlit app
def run_streamlit_app(app_path, port=8501):
    """
    Run Streamlit app in a separate process
    """
    cmd = f"streamlit run {app_path} --server.port {port} --server.headless true"
    
    try:
        print(f"🚀 Starting Streamlit app...")
        print(f"📱 Command: {cmd}")
        print(f"🌐 Local URL: http://localhost:{port}")
        print("\n⏳ Starting server (this may take a few moments)...")
        
        # Start the process
        process = subprocess.Popen(
            cmd.split(),
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True,
            cwd=str(project_root)
        )
        
        return process
        
    except Exception as e:
        print(f"❌ Error starting Streamlit: {e}")
        return None

# Test if we can import streamlit
try:
    import streamlit as st
    print("✅ Streamlit is available")
    print(f"📦 Streamlit version: {st.__version__}")
    
    # You can start the app manually by running this cell
    print("\n🔧 To start the app manually, run:")
    print(f"streamlit run {app_file} --server.port 8501")
    
except ImportError:
    print("❌ Streamlit not installed. Install it with: pip install streamlit")
````

## 4. 🌐 Ngrok Setup for Public Access

Ngrok allows us to create a public tunnel to our local Streamlit app.

````python
# Install and setup ngrok
try:
    from pyngrok import ngrok, conf
    print("✅ pyngrok is available")
    
    # Check if ngrok is installed
    try:
        ngrok_path = conf.get_default().ngrok_path
        print(f"📍 Ngrok path: {ngrok_path}")
    except Exception as e:
        print(f"⚠️ Ngrok binary not found: {e}")
        print("\n📥 Installing ngrok...")
        
        # Install ngrok binary
        conf.get_default().ngrok_path = None
        
except ImportError:
    print("❌ pyngrok not installed. Install it with: pip install pyngrok")
    
# Function to create ngrok tunnel
def create_ngrok_tunnel(port=8501):
    """
    Create a public ngrok tunnel to the Streamlit app
    """
    try:
        from pyngrok import ngrok
        
        # Create tunnel
        print(f"🔗 Creating ngrok tunnel for port {port}...")
        public_tunnel = ngrok.connect(port)
        
        print(f"\n🎉 Ngrok tunnel created successfully!")
        print(f"🌐 Public URL: {public_tunnel.public_url}")
        print(f"🏠 Local URL: http://localhost:{port}")
        
        return public_tunnel
        
    except Exception as e:
        print(f"❌ Error creating ngrok tunnel: {e}")
        return None

print("🔧 Ngrok setup functions ready")
````

## 5. 🎯 Complete Deployment Process

Here's the complete process to deploy your heart disease prediction app:

````python
# Complete deployment function
def deploy_heart_disease_app():
    """
    Complete deployment process for the heart disease prediction app
    """
    print("🚀 Starting Heart Disease Prediction App Deployment\n")
    
    # Step 1: Verify app file exists
    app_file = project_root / 'ui' / 'app.py'
    if not app_file.exists():
        print(f"❌ App file not found: {app_file}")
        return
    
    print(f"✅ App file found: {app_file}")
    
    # Step 2: Check dependencies
    required_packages = ['streamlit', 'plotly', 'pandas', 'numpy', 'scikit-learn']
    missing = []
    
    for pkg in required_packages:
        try:
            __import__(pkg.replace('-', '_'))
        except ImportError:
            missing.append(pkg)
    
    if missing:
        print(f"❌ Missing packages: {missing}")
        print(f"📦 Install with: pip install {' '.join(missing)}")
        return
    
    print("✅ All dependencies satisfied")
    
    # Step 3: Instructions for manual deployment
    print("\n📋 Deployment Instructions:")
    print("="*50)
    
    print("\n1. 🏠 Local Deployment:")
    print(f"   streamlit run {app_file}")
    print("   Access at: http://localhost:8501")
    
    print("\n2. 🌐 Public Deployment with Ngrok:")
    print("   # Terminal 1: Start Streamlit")
    print(f"   streamlit run {app_file}")
    print("   ")
    print("   # Terminal 2: Start Ngrok")
    print("   ngrok http 8501")
    
    print("\n3. 🔧 Alternative Python Deployment:")
    print("   # Run from this notebook:")
    print("   import subprocess")
    print(f"   subprocess.Popen(['streamlit', 'run', '{app_file}'])")
    
    print("\n🎯 App Features:")
    print("• Interactive patient data input")
    print("• Multiple ML model selection")
    print("• Real-time risk prediction")
    print("• Feature importance visualization")
    print("• Professional medical UI")
    
    print("\n⚠️ Important Notes:")
    print("• This is for educational purposes only")
    print("• Always consult healthcare professionals")
    print("• Keep the terminal running while using the app")
    
    return True

# Run deployment process
deploy_heart_disease_app()
````

## 6. 📊 Deployment Testing and Monitoring

Let's create some utilities to test and monitor our deployed application.

````python
import requests
import json
from datetime import datetime

def test_app_endpoint(url, timeout=10):
    """
    Test if the Streamlit app is accessible
    """
    try:
        print(f"🔍 Testing app at: {url}")
        response = requests.get(url, timeout=timeout)
        
        if response.status_code == 200:
            print(f"✅ App is accessible (Status: {response.status_code})")
            print(f"📏 Response size: {len(response.content)} bytes")
            return True
        else:
            print(f"⚠️ App responded with status: {response.status_code}")
            return False
            
    except requests.exceptions.ConnectionError:
        print(f"❌ Cannot connect to {url}")
        print("   Make sure the Streamlit app is running")
        return False
    except requests.exceptions.Timeout:
        print(f"⏰ Request timed out after {timeout} seconds")
        return False
    except Exception as e:
        print(f"❌ Error testing endpoint: {e}")
        return False

def generate_deployment_report():
    """
    Generate a comprehensive deployment report
    """
    report = {
        'timestamp': datetime.now().isoformat(),
        'project_name': 'Heart Disease Prediction App',
        'status': 'Ready for Deployment',
        'files': {
            'app_file': str(project_root / 'ui' / 'app.py'),
            'config_file': str(project_root / '.streamlit' / 'config.toml'),
            'requirements': str(project_root / 'requirements.txt')
        },
        'deployment_options': [
            'Local (localhost:8501)',
            'Ngrok Tunnel (public URL)',
            'Streamlit Cloud',
            'Heroku',
            'AWS/GCP/Azure'
        ],
        'app_features': [
            'Interactive patient data input',
            'Multiple ML model selection',
            'Real-time predictions',
            'Feature importance analysis',
            'Professional medical interface',
            'Risk assessment visualization'
        ]
    }
    
    print("📊 Deployment Report")
    print("="*50)
    print(f"🕐 Generated: {report['timestamp']}")
    print(f"📱 Project: {report['project_name']}")
    print(f"🚦 Status: {report['status']}")
    
    print("\n📁 Key Files:")
    for name, path in report['files'].items():
        exists = "✅" if Path(path).exists() else "❌"
        print(f"  {exists} {name}: {path}")
    
    print("\n🚀 Deployment Options:")
    for i, option in enumerate(report['deployment_options'], 1):
        print(f"  {i}. {option}")
    
    print("\n⭐ App Features:")
    for feature in report['app_features']:
        print(f"  • {feature}")
    
    # Save report to file
    report_file = project_root / 'deployment' / 'deployment_report.json'
    report_file.parent.mkdir(exist_ok=True)
    
    with open(report_file, 'w') as f:
        json.dump(report, f, indent=2)
    
    print(f"\n💾 Report saved to: {report_file}")
    
    return report

# Generate the deployment report
deployment_report = generate_deployment_report()
````
